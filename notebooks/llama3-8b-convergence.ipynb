```python
import fsspec

fs = fsspec.filesystem('gcs')


def load_files(name):
  path = f'gs://torchprime/yifeit-exp/{name}/outputs/0-0/debug_dataloader/'
  files = fs.ls(path)
  files = sorted(list(files))
  print("Files:", files)
  assert len(files) > 0, "No files found in the specified GCS path."
  return files

hf_files = load_files('hf-dump-inputs')
tp_files = load_files('tp-llama-3-8b-dump-inputs-2')
```

```python
def load_input(path):
  with fs.open(path, 'rb') as f:
    content = f.read()
  import json
  obj = json.loads(content)
  assert isinstance(obj, dict), "Loaded content is not a dictionary."
  return obj
```

```python
hf_input = load_input(hf_files[0])
print(hf_input.keys())
```

```python
tp_input = load_input(tp_files[0])
print(tp_input.keys())
```

# Compare data loaded in the first training step

```python
assert hf_input['input_ids'] == tp_input['input_ids']
assert hf_input['attention_mask'] == tp_input['attention_mask']
assert hf_input['labels'] == tp_input['labels']
```
